This notebook explores the Titanic passenger dataset using Pandas. It covers DataFrame inspection, missing-value handling, duplicate checks, grouping, aggregation, and basic insights.

**Dataset:** Titanic CSV (public GitHub mirror).

## 1. Import libraries and load dataset
Run this notebook in Google Colab or Jupyter with an internet connection.

```python
import pandas as pd
import matplotlib.pyplot as plt

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)
print('Dataset loaded successfully!')
print('Shape:', df.shape)
df.head()
```

## 2. DataFrame operations: head(), info(), describe()

```python
print('First five rows:')
display(df.head())

print('Dataset information:')
df.info()

print('Statistical summary:')
display(df.describe(include='all'))
```

## 3. Missing values and duplicate records
First inspect missing values and duplicate rows. Then fill selected missing values using median/mode and remove duplicate rows if any.

```python
print('Missing values before cleaning:')
display(df.isnull().sum())
print('Duplicate rows before cleaning:', df.duplicated().sum())

# Fill missing numerical age values with the median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked values with the most frequent category
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Cabin has many missing values, so create a simple indicator instead of guessing cabin numbers
df['HasCabin'] = df['Cabin'].notna().astype(int)

# Remove duplicate rows, if present
df = df.drop_duplicates().copy()

print('Missing values after selected cleaning:')
display(df.isnull().sum())
print('Duplicate rows after cleaning:', df.duplicated().sum())
```

## 4. Grouping and aggregation analysis

```python
# Survival count and survival rate
survival_summary = df.groupby('Survived').agg(
    passenger_count=('PassengerId', 'count'),
    average_age=('Age', 'mean'),
    average_fare=('Fare', 'mean')
)
display(survival_summary)

# Survival rate by passenger class
class_summary = df.groupby('Pclass').agg(
    passengers=('PassengerId', 'count'),
    survival_rate=('Survived', 'mean'),
    average_fare=('Fare', 'mean')
)
display(class_summary)

# Survival rate by gender
gender_summary = df.groupby('Sex').agg(
    passengers=('PassengerId', 'count'),
    survival_rate=('Survived', 'mean')
)
display(gender_summary)
```

## 5. Visualize key findings

```python
df.groupby('Pclass')['Survived'].mean().plot(kind='bar', ylim=(0, 1),
    title='Survival Rate by Passenger Class', ylabel='Survival rate', xlabel='Passenger class')
plt.tight_layout()
plt.show()

df.groupby('Sex')['Survived'].mean().plot(kind='bar', ylim=(0, 1),
    title='Survival Rate by Gender', ylabel='Survival rate', xlabel='Gender')
plt.tight_layout()
plt.show()
```
